### List Indexes

````python
colors = ['red', 'green', 'blue', 'yellow', 'white', 'black']
````

````python
# Notice the value in the 3rd position (index 2) has changed.
colors
````

## 🏋🏽 Doing things to lists: Functions/Methods

````python
pollutants = ['co', 'co2', 'no2', 'o3']
````

````python
# Finding the number of items in a list
len(pollutants)
````

````python
# Appending a new value
pollutants.append('ch4')
````

````python
# Updated list
print(pollutants)
````

````python
# sort a list alphabetically (forward and reverse)
pollutants.sort()

print(pollutants)
````

Summary of Big Ideas
1. Lists are a data type that hold a group of values.  They are defined with square brackets `[]` with commas in between values.
2. **Indexes** are ways to access items in your list using the position of the value in the list as a number.
2. In python we start counting from 0, not from 1 with indexes.
2. Functions/methods (think of those two as the same thing right now) are actions you can take that let you ask a question about an object or change it somehow.  The two we looked at today are `len()` and `.append()`, which demonstrate the two common types of syntax.

### Practice (Challenge)

Print out the following pieces of information from the dictionary above:

- the motto
- the population ages 25 to 50
- Change the population ages 0 to 25 to 98988
- Add a new age bracket to the population dictionary by adding an additional item: ages_50_75 with the population 3333. Change the item ages_over_50 to be renamed ages_over_75.

````python
state_info = {
    'state': 'AK',
    'motto': 'north to the future',
    'num_of_counties': 29,
    'population': {
        'ages_0_25': 28393,
        'ages_25_50': 25920,
        'ages_over_50': 8943,
        }
}
````

````python
# motto
state_info['motto']

# the population ages 25 to 50
state_info['population']['ages_25_50']

# Change the population ages 0 to 25 to 98988
state_info['population']['ages_25_50'] = 98988

# Add a new age bracket / change age bracket
state_info['population']['ages_50_75'] = 3333
state_info['population']['ages_over_75'] = state_info['population'].pop('ages_over_50')
````

````python
print(state_info)
````
