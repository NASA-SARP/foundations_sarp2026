### Practice (Challenge): Nested Dictionary

Goal of the code snippet: This code takes a set of chlorophyll observations and extracts the sea surface value of a few target locations. It checks to see if the observed sst is the 25 degrees Celsius, and prints if it has or has not located a point with the target temperature.

_Hint: We haven't looked at the `.get()` method together. Read more about `.get()` [here](https://www.programiz.com/python-programming/methods/dictionary/get)_

```python
ocean_obs = {
    'point1': {'sst': 25, 'chlorophyll': 0.8, 'location': [28.84, -95.73]},
    'point2': {'sst': 24, 'chlorophyll': 2, 'location': [28.80, -95.73]},
    'point3': {'sst': 26, 'chlorophyll': 1.5, 'location': [27.99, -96.00]},
    'point4': {'sst': 25, 'chlorophyll': 2.1, 'location': [28.23, -96.03]},
}

point2_temp = ocean_obs['point2'].get('temp')
point4_temp = ocean_obs['point4'].get('temp')

print('Checking Point 2')
if point2_temp == 25:
    print('located target temperature')

print('Checking Point 4')
if point4_temp == 25:
    print('located target temperature')
```
